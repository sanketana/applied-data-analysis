## Part 1 — Quick recap: if/else and loops on single values

### Classifying a patient's blood pressure

````python
systolic_bp = 135

if systolic_bp < 120:
    category = "Normal"
elif systolic_bp < 130:
    category = "Elevated"
else:
    category = "High"

print(f"BP {systolic_bp} → {category}")
````

### Looping through a list of values

````python
bp_values = [119, 135, 128, 142, 115]

for bp in bp_values:
    if bp < 120:
        print(f"{bp} → Normal")
    elif bp < 130:
        print(f"{bp} → Elevated")
    else:
        print(f"{bp} → High")
````

This works for a small list. But what if we have 20 patients in a CSV? Or 9,000 in NHANES? We need a better way.

## Part 3 — Filtering and computing stats

Now combine filtering with the stats you already know.

### Filter by sex

````python
females = df[df['sex'] == 'F']
print(f"Female patients: {len(females)}")
females.head()
````

### Filter + stat in one line

Remember the Session 2 stretch exercise? Here's why it works:

````python
df[df['sex'] == 'M']['weight_kg'].mean()
````

Read it left to right:
1. `df['sex'] == 'M'` → boolean Series (True for males)
2. `df[...]` → filter to only male rows
3. `['weight_kg']` → select the weight column
4. `.mean()` → compute the average

Each step feeds into the next.

### Group comparison: mean systolic BP by sex

````python
mean_bp_male = df[df['sex'] == 'M']['systolic_bp'].mean()
mean_bp_female = df[df['sex'] == 'F']['systolic_bp'].mean()

print(f"Mean systolic BP — Males: {mean_bp_male:.1f} mmHg")
print(f"Mean systolic BP — Females: {mean_bp_female:.1f} mmHg")
````

### Combining conditions with & (and) and | (or)

To filter on two conditions at once, use `&` for "and" and `|` for "or". Each condition must be wrapped in parentheses.

````python
older_females = df[(df['age'] > 40) & (df['sex'] == 'F')]
print(f"Females over 40: {len(older_females)}")
older_females
````

**Common mistake:** forgetting the parentheses. Try running this and read the error:

```python
df[df['age'] > 40 & df['sex'] == 'F']   # this will break
```

The fix is always: put parentheses around each condition.

### High blood pressure patients

````python
high_bp = df[df['systolic_bp'] >= 130]
print(f"Patients with high BP (≥130): {len(high_bp)}")
print(f"Their mean age: {high_bp['age'].mean():.1f} years")
````

## Part 4 — The stall point: loop vs pandas

**Task:** compute the mean weight of patients over 50.

### Approach 1: a for loop

````python
total = 0
count = 0
for i in range(len(df)):
    if df['age'][i] > 50:
        total += df['weight_kg'][i]
        count += 1

print(f"Loop result: {total / count:.1f} kg")
````

### Approach 2: pandas boolean filtering

````python
result = df[df['age'] > 50]['weight_kg'].mean()
print(f"Pandas result: {result:.1f} kg")
````

Same answer. Six lines vs one. With 9,000 NHANES rows, the loop would be noticeably slower.

**The rule:** if you're writing a for loop over DataFrame rows, there's almost always a pandas way to do it instead.

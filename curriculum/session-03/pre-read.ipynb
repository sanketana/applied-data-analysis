## Comparison operators

You already know `=` assigns a value to a variable. Python uses different symbols to *compare* values:

````python
age = 45

print(age == 45)    # equal to
print(age != 30)    # not equal to
print(age > 40)     # greater than
print(age < 40)     # less than
print(age >= 45)    # greater than or equal to
print(age <= 50)    # less than or equal to
````

Each comparison produces `True` or `False`. These are **boolean values** — the same type you saw in Session 1.

Notice the double equals `==` for comparison. A single `=` assigns; a double `==` compares. This trips up everyone at least once.

## if / elif / else

Comparisons become powerful when you use them to make decisions. Here's the pattern:

````python
systolic_bp = 135

if systolic_bp < 120:
    print("Normal blood pressure")
elif systolic_bp < 130:
    print("Elevated blood pressure")
else:
    print("High blood pressure")
````

How to read this:

1. Python checks the `if` condition first. If it's `True`, it runs that indented block and skips everything else.
2. If the `if` was `False`, it checks `elif` ("else if"). You can have as many `elif` branches as you need.
3. If nothing above matched, the `else` block runs. It's the catch-all.

**The indentation matters.** The indented lines under each branch are what runs when that condition is true. Python uses indentation (4 spaces) to know which code belongs to which branch.

Here's another example — classifying a patient's age group:

````python
patient_age = 67

if patient_age < 18:
    group = "minor"
elif patient_age < 65:
    group = "adult"
else:
    group = "senior"

print(f"Age {patient_age} → {group}")
````

## Boolean operators: and, or, not

Sometimes you need to check more than one condition at once.

````python
age = 45
systolic_bp = 135

# Both must be true
if age > 40 and systolic_bp > 130:
    print("Over 40 with high BP — flag for follow-up")

# At least one must be true
if age > 65 or systolic_bp > 140:
    print("At least one risk factor present")

# Flip the result
has_diabetes = False
if not has_diabetes:
    print("No diabetes recorded")
````

## for loops

A `for` loop repeats an action for each item in a list. Instead of writing the same code five times, you write it once and let Python repeat it.

````python
ages = [34, 45, 62, 28, 71]

for age in ages:
    print(f"Patient age: {age}")
````

How it works:

1. Python picks the first item from `ages` (34), assigns it to `age`, and runs the indented block.
2. Then it picks the next item (45), assigns it to `age`, and runs the block again.
3. This continues until every item has been processed.

The variable name `age` is your choice — it could be `x` or `patient_age` or anything. But a descriptive name makes the code easier to read.

### Combining loops with if/else

This is where it gets useful. You can put an if/else *inside* a loop to classify each item:

````python
ages = [34, 45, 62, 28, 71]

for age in ages:
    if age >= 65:
        print(f"{age} → senior")
    else:
        print(f"{age} → not senior")
````

Notice the indentation: the `if/else` is indented once (inside the loop), and the `print` statements are indented twice (inside the if/else, which is inside the loop).

### Building up a result with a loop

A common pattern: start with a counter or total, then update it inside the loop.

````python
ages = [34, 45, 62, 28, 71]

count_over_50 = 0
for age in ages:
    if age > 50:
        count_over_50 += 1

print(f"{count_over_50} patients are over 50")
````

The `+=` operator means "add to the current value." So `count_over_50 += 1` is the same as `count_over_50 = count_over_50 + 1`.

### range()

`range()` generates a sequence of numbers. It's useful when you need to repeat something a specific number of times, or when you need the index (position) of each item.

````python
# Print numbers 0 through 4
for i in range(5):
    print(i)
````

````python
# range(start, stop) — stop is NOT included
for i in range(1, 4):
    print(i)
````

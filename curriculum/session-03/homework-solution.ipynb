## Part 1 — if/else classification practice

The learner picks three patients, computes BMI, and classifies each.

````python
# Patient P001: height 176.3 cm, weight 79.5 kg
height_m = 176.3 / 100
weight = 79.5
bmi = weight / (height_m ** 2)

if bmi < 18.5:
    cat = "Underweight"
elif bmi < 25:
    cat = "Normal"
elif bmi < 30:
    cat = "Overweight"
else:
    cat = "Obese"

print(f"Patient P001: BMI = {bmi:.1f} → {cat}")

# Patient P006: height 159.0 cm, weight 73.2 kg
height_m = 159.0 / 100
weight = 73.2
bmi = weight / (height_m ** 2)

if bmi < 18.5:
    cat = "Underweight"
elif bmi < 25:
    cat = "Normal"
elif bmi < 30:
    cat = "Overweight"
else:
    cat = "Obese"

print(f"Patient P006: BMI = {bmi:.1f} → {cat}")

# Patient P015: height 167.6 cm, weight 47.7 kg
height_m = 167.6 / 100
weight = 47.7
bmi = weight / (height_m ** 2)

if bmi < 18.5:
    cat = "Underweight"
elif bmi < 25:
    cat = "Normal"
elif bmi < 30:
    cat = "Overweight"
else:
    cat = "Obese"

print(f"Patient P015: BMI = {bmi:.1f} → {cat}")
````

**Expected output (with these patients):**
```
Patient P001: BMI = 25.6 → Overweight
Patient P006: BMI = 29.0 → Overweight
Patient P015: BMI = 17.0 → Underweight
```

**Instructor notes:**
- Learner picks different patients, so values will differ. Check that the BMI is physiologically reasonable and the classification matches.
- Common mistake: forgetting to convert height from cm to m. If BMIs are tiny (like 0.25), that's the cause.
- Some learners will copy-paste the if/elif block three times. That's fine — they haven't learned functions yet (Session 4). If they ask "is there a way to avoid repeating this?", say: "Yes, that's exactly what Session 4 is about."
- The boundary case `bmi < 25` vs `bmi < 25.0` doesn't matter in practice, but if a learner asks, acknowledge it.

## Part 2 — For loop practice

The learner loops through 5 systolic BP values and classifies each.

````python
bp_values = [119, 137, 128, 115, 143]

for bp in bp_values:
    if bp < 120:
        print(f"{bp} → Normal")
    elif bp < 130:
        print(f"{bp} → Elevated")
    else:
        print(f"{bp} → High")
````

**Expected output (with these values):**
```
119 → Normal
137 → High
128 → Elevated
115 → Normal
143 → High
```

**Instructor notes:**
- Values will differ. Check that all values come from the CSV (range 110–149) and classifications are correct.
- If the learner wrote separate if statements instead of if/elif/else (checking each range independently), it still works but isn't clean. Point out the difference: with elif, only one branch runs.

## Part 3 — Boolean filtering on the patient CSV

````python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd

csv_path = "/content/drive/MyDrive/LEARNER-FOLDER/session-01/patient_measurements_small.csv"
df = pd.read_csv(csv_path)
````

````python
# 3a. Patients over 40
over_40 = df[df['age'] > 40]
print(f"Patients over 40: {len(over_40)}")
````

````python
# 3b. Female patients
females = df[df['sex'] == 'F']
print(f"Female patients: {len(females)}")
females
````

````python
# 3c. Mean systolic BP by sex
print(f"Mean systolic BP — Males: {df[df['sex'] == 'M']['systolic_bp'].mean():.1f} mmHg")
print(f"Mean systolic BP — Females: {df[df['sex'] == 'F']['systolic_bp'].mean():.1f} mmHg")
````

````python
# 3d. Patients with systolic BP > 130
high_bp = df[df['systolic_bp'] > 130]
print(f"Patients with BP > 130: {len(high_bp)}")
print(f"Their mean age: {high_bp['age'].mean():.1f} years")
````

````python
# 3e. Females over 50
f_over_50 = df[(df['sex'] == 'F') & (df['age'] > 50)]
print(f"Females over 50: {len(f_over_50)}")
print(f"Their mean weight: {f_over_50['weight_kg'].mean():.1f} kg")
````

**Expected outputs (from the 20-row patient CSV):**
- 3a: 14 patients over 40
- 3b: 10 female patients
- 3c: Males ~126.3 mmHg, Females ~132.4 mmHg
- 3d: 11 patients with BP > 130, mean age ~52.5 years
- 3e: 5 females over 50, mean weight ~67.3 kg

**Instructor notes:**
- 3e is the first time the learner combines two conditions with `&`. If they got stuck on the parentheses, that's expected. Check if they found the fix.
- If the learner used `and` instead of `&` and got a ValueError, that's the exact mistake the classwork warned about. Good learning moment.

## Part 5 — NHANES demographics filtering

````python
demo = pd.read_sas("/content/drive/MyDrive/LEARNER-FOLDER/session-02/DEMO_J.XPT")
````

````python
# 5a. Adults only
adults = demo[demo['RIDAGEYR'] >= 18]
print(f"Adults in NHANES: {len(adults)}")
````

````python
# 5b. Count by gender
n_male = (demo['RIAGENDR'] == 1).sum()
n_female = (demo['RIAGENDR'] == 2).sum()
print(f"Males: {n_male}")
print(f"Females: {n_female}")
````

````python
# 5c. Mean age of adult males vs adult females
adult_males = demo[(demo['RIDAGEYR'] >= 18) & (demo['RIAGENDR'] == 1)]
adult_females = demo[(demo['RIDAGEYR'] >= 18) & (demo['RIAGENDR'] == 2)]

print(f"Mean age of adult males: {adult_males['RIDAGEYR'].mean():.1f} years")
print(f"Mean age of adult females: {adult_females['RIDAGEYR'].mean():.1f} years")
````

````python
# 5d. Summary sentence
print(f"The NHANES 2017-2018 survey includes {len(adults)} adults. Mean age: males {adult_males['RIDAGEYR'].mean():.1f} years, females {adult_females['RIDAGEYR'].mean():.1f} years.")
````

**Expected output (approximate, from NHANES 2017-2018):**
- 5a: around 6,150 adults
- 5b: around 4,550 males, 4,700 females (total survey, not just adults)
- 5c: Mean age of adult males ~47 years, adult females ~48 years
- 5d: Summary with the numbers above

**Instructor notes:**
- 5c is the most challenging question because it requires combining two conditions with `&`. If the learner got this right independently, they've internalized boolean filtering.
- Common mistake: using `demo['RIAGENDR'] == 'M'` (string) instead of `== 1` (numeric). NHANES codes gender as 1/2, not M/F. If the learner hit this, it's a data-awareness lesson: always check what the values actually look like before filtering.
- The age cap at 80 in NHANES means the mean age for seniors is slightly depressed. If the learner notices this, excellent observation.

## Part 6 — Reflection

Sample answers (learner responses will vary).

**Sample answers:**

1. *What does `df[df['age'] > 50]` do?* First, `df['age'] > 50` compares every value in the age column to 50 and produces a True/False value for each row. Then `df[...]` uses that True/False column to pick out only the rows where the result is True. You get back a smaller DataFrame with only the patients over 50.

2. *Why is pandas better than a loop?* The pandas way is shorter (one line vs many), easier to read, and faster on large datasets. With a loop, you have to manage counters and totals yourself. Pandas handles all of that internally.

3. Varies. Good answers reference boolean filtering, the `&` vs `and` gotcha, or the "don't loop through DataFrames" rule.

**Instructor notes:**
- For question 1, the key is that the learner can describe the two-step process: create the boolean Series, then use it to filter. If they just say "it gives me patients over 50" without mentioning the boolean step, push for more detail.
- If the reflection is thin, review in Session 4 warm-up.

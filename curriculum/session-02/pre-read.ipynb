## Part 1 — Lists: ordered collections


A **list** is Python's way of holding several values in a specific order. Square brackets, comma-separated.

Think of it like a row of measurements you wrote down.


```python
# A list of ages from a clinical study
ages = [34, 45, 62, 28, 51]
print(ages)
```

### Accessing elements: indexing

Lists are **zero-indexed** — the first element is at position 0, not 1. This trips up almost everyone at first. Repeat after me: **zero is the first position**.


```python
print(ages[0])   # first element
print(ages[1])   # second element
print(ages[-1])  # last element (negative indices count from the end)
```

### Basic operations on lists


```python
print(len(ages))     # how many elements?
print(sum(ages))     # add them all up
print(min(ages))     # smallest
print(max(ages))     # largest
print(sum(ages) / len(ages))   # mean!
```

Pause here. Re-read that last line. You just computed the **mean age** of a group of patients using only Python's built-in functions. No pandas required.

This is the foundation. Pandas does the same thing, just on big data, with prettier syntax.


### Adding to a list


```python
ages.append(39)    # add one value to the end
print(ages)
```

### Slicing — grabbing a range


```python
print(ages[0:3])   # first three elements (positions 0, 1, 2)
print(ages[2:])    # from position 2 to the end
print(ages[:3])    # from the start up to (not including) position 3
```

**The slicing rule:** `ages[start:stop]` gives you elements from `start` up to *but not including* `stop`. So `ages[0:3]` gives positions 0, 1, and 2 — three elements.

## Part 2 — Dictionaries: labeled values


A **dictionary** holds values associated with labels (called **keys**).

Think of it like one patient's record card: name, age, sex, weight — each with a label.


```python
patient = {
    "patient_id": "P001",
    "age": 58,
    "sex": "M",
    "height_cm": 176.3,
    "weight_kg": 79.5
}
print(patient)
```

### Accessing values by key


```python
print(patient["age"])
print(patient["height_cm"])
```

Notice: with a **list**, you access by position (`ages[0]`). With a **dictionary**, you access by key name (`patient["age"]`). This difference matters.


### Updating and adding


```python
patient["age"] = 59   # update existing value
patient["has_diabetes"] = False   # add new key-value pair
print(patient)
```

### Checking keys


```python
print(list(patient.keys()))     # all the keys
print(list(patient.values()))   # all the values
```

## Part 3 — Why these two shapes matter


In data analysis, you'll constantly work with tabular data — rows and columns. Here's the key insight:

**A table can be seen two ways:**

1. As a **list of dictionaries** — each row is a dictionary (one patient's record), and the table is a list of these.
2. As a **dictionary of lists** — each column is a list (all the ages, all the weights), and the table is a dictionary of these columns.

Both views are valid. Pandas leans toward view #2 (dictionary of lists) because most analysis operates on columns. But you'll use both mental models throughout this course.


```python
# A tiny table, expressed as a list of patient-dictionaries
patients = [
    {"id": "P001", "age": 58, "weight_kg": 79.5},
    {"id": "P002", "age": 65, "weight_kg": 55.3},
    {"id": "P003", "age": 45, "weight_kg": 79.6},
]

# Access the second patient's age
print(patients[1]["age"])   # position 1 in the list, then key 'age'
```

```python
# The same data, expressed as a dictionary of columns
columns = {
    "id": ["P001", "P002", "P003"],
    "age": [58, 65, 45],
    "weight_kg": [79.5, 55.3, 79.6],
}

# Access the second patient's age (different route)
print(columns["age"][1])   # the 'age' column, then position 1
```

Sit with this for a moment. **Both pieces of code give the same answer, but the data is organized differently.** This is the mental model behind pandas — we'll build on it in class.

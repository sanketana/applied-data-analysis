## Part 3 — Selecting columns

### One column — single brackets → a Series

```python
df['age']
```

```python
df['age'].mean()
```

```python
df['age'].max()
```

Pause. This is the moment: `df['age']` behaves like a list, but with powers (`.mean()`, `.max()`, and many more). **A column is a list with superpowers.** Add that to your reference sheet.

### Multiple columns — double brackets → a DataFrame

```python
df[['age', 'weight_kg']]
```

### The single-vs-double bracket distinction

```python
single = df['age']
double = df[['age']]

print(type(single))
print(type(double))
```

These look almost identical but pandas tells us they're different types. Why might that matter?

*(Discussion with instructor.)*

### The rule to write in your reference sheet now

- `df['col']` → Series (one column, list-like)
- `df[['col1', 'col2']]` → DataFrame (one or more columns, still a table)
- For multiple columns, the **list of column names** goes inside another set of brackets.

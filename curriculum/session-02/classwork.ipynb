## Part 3 — Selecting columns

### One column — single brackets → a Series

```python
df['age']
```

```python
df['age'].mean()
```

```python
df['age'].max()
```

Pause. This is the moment: `df['age']` behaves like a list, but with powers (`.mean()`, `.max()`, and many more). **A column is a list with superpowers.** Add that to your reference sheet.

### Multiple columns — double brackets → a DataFrame

```python
df[['age', 'weight_kg']]
```

### The single-vs-double bracket distinction

```python
single = df['age']
double = df[['age']]

print(type(single))
print(type(double))
```

These look almost identical but pandas tells us they're different types. Why might that matter?

*(Discussion with instructor.)*

### The rule to write in your reference sheet now

- `df['col']` → Series (one column, list-like)
- `df[['col1', 'col2']]` → DataFrame (one or more columns, still a table)
- For multiple columns, the **list of column names** goes inside another set of brackets.

## Part 5 — First real-world dataset: NHANES

So far we've been working with a small, made-up patient CSV. Now let's use **real data** from a real national health survey.

**NHANES** (National Health and Nutrition Examination Survey) is run by the CDC. Every two years, they examine thousands of Americans: blood tests, body measurements, questionnaires about diet, exercise, and more. All the data is free and public.

We'll use NHANES data throughout this course. Today is your first taste.

### Step 1 — Download the demographics file from the CDC website

Follow these steps:

1. Go to the NHANES data page: **https://wwwn.cdc.gov/nchs/nhanes/Default.aspx**
2. Under "Survey Years", click **2017-2018**
3. Click **Demographics Data** in the list
4. Find the file called **Demographic Variables and Sample Weights (DEMO_J)** and click the **Data File** link (it says "XPT" with the file size next to it)
5. A file called `DEMO_J.XPT` will download to your computer

**What's an XPT file?** It's a data format used by SAS (a statistical software package). NHANES distributes all its data in this format. The good news: pandas can read it directly.

### Step 2 — Upload to your Google Drive

1. Open your Google Drive
2. Navigate to your course folder, then into `/session-02/`
3. Drag and drop `DEMO_J.XPT` into that folder
4. Wait for the upload to finish (the file is a few MB, should take seconds)

### Step 3 — Read the XPT file with pandas

For CSV files we used `pd.read_csv()`. For XPT files, we use `pd.read_sas()`.

⚠️ Update the path below to match your folder.

```python
demo = pd.read_sas("/content/drive/MyDrive/YOUR-FOLDER-HERE/session-02/DEMO_J.XPT")
demo.head()
```

### Step 4 — Inspect it: the same four questions

```python
demo.dtypes
```

### Understanding the column names

NHANES uses short, coded column names. Here are the ones we'll care about today:

| Column | What it means |
|---|---|
| `SEQN` | Respondent sequence number (unique ID for each person) |
| `RIAGENDR` | Gender (1 = Male, 2 = Female) |
| `RIDAGEYR` | Age in years at screening |
| `RIDRETH3` | Race/Hispanic origin |
| `DMDEDUC2` | Education level (adults 20+) |
| `INDHHIN2` | Annual household income |

You can look up any NHANES variable in the online codebook. For now, just know that every column has a documented meaning.

### Step 5 — Select columns and compute stats

Let's use the exact same pandas skills from Parts 3 and 4, but on real data.

```python
# Age column
demo['RIDAGEYR']
```

```python
print(f"Mean age: {demo['RIDAGEYR'].mean():.1f} years")
print(f"Youngest: {demo['RIDAGEYR'].min():.0f} years")
print(f"Oldest: {demo['RIDAGEYR'].max():.0f} years")
```

**Expected output:**
- Mean age: around 31 years (NHANES includes children)
- Youngest: 0 (infants are included in the survey)
- Oldest: 80 (NHANES caps reported age at 80)

```python
# Select multiple columns — same double-bracket rule
demo[['SEQN', 'RIDAGEYR', 'RIAGENDR']].head(10)
```

```python
# How many people in the survey?
print(f"Total respondents: {len(demo)}")
```

### What just happened

You loaded a real dataset from a US government health survey and computed statistics on it. The pandas skills are exactly the same as with the 20-row patient CSV:

- `pd.read_sas()` instead of `pd.read_csv()` (different file format, same idea)
- `.head()`, `.shape`, `.columns`, `.dtypes` to inspect
- Single brackets for one column, double brackets for multiple
- `.mean()`, `.min()`, `.max()` for stats

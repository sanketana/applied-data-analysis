## Block D — Single vs double brackets (stall point)

The template walks through this step-by-step. Key expected outputs:

```python
single = df['age']
double = df[['age']]

print(type(single))   # <class 'pandas.core.series.Series'>
print(type(double))   # <class 'pandas.core.frame.DataFrame'>
```

```python
# Both work but return different shapes
print(single.mean())   # 48.1 (a scalar)
print(double.mean())   # age    48.1 (a Series with column name)
```

```python
# Multi-column selection — correct way
df[['age', 'weight_kg']]
```

```python
# The classic error — uncomment to demonstrate
# df['age', 'weight_kg']   # KeyError: ('age', 'weight_kg')
```

**Instructor notes:**
- The error from `df['age', 'weight_kg']` is a `KeyError` because pandas interprets the comma as a tuple key, not two separate columns.
- If the learner asks "why did pandas design it this way?" — honest answer: it's historical. Single brackets were for Series access (like a dict), double brackets are syntactic sugar for passing a list of column names. Both are useful; neither is a mistake.
- The rule for the reference sheet: single brackets = Series, double brackets = DataFrame. For multiple columns, the list of names goes inside another set of brackets.

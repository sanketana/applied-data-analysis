## Part 1 — List practice with real data

The learner picks five systolic BP values from the CSV and computes stats using plain Python.

````python
systolic_bps = [119, 110, 137, 131, 115]

# 1. Mean
mean_bp = sum(systolic_bps) / len(systolic_bps)
print(f"Mean systolic BP: {mean_bp:.1f} mmHg")

# 2. Highest and lowest
print(f"Highest: {max(systolic_bps)} mmHg")
print(f"Lowest: {min(systolic_bps)} mmHg")

# 3. Range
bp_range = max(systolic_bps) - min(systolic_bps)
print(f"Range: {bp_range} mmHg")

# 4. Classification message
print(f"The range of systolic BP in my sample is {bp_range} mmHg")
````

**Expected output (with these five values):**
```
Mean systolic BP: 122.4 mmHg
Highest: 137 mmHg
Lowest: 110 mmHg
Range: 27 mmHg
The range of systolic BP in my sample is 27 mmHg
```

**Instructor notes:**
- Values will differ depending on which five patients the learner picked. Check that the values are actual systolic BPs from the CSV (range 110–149 in this dataset).
- Common mistake: computing `sum(systolic_bps) / 5` instead of `/ len(systolic_bps)`. Works here, but gently point out that `len()` is more robust.
- If the learner didn't use an f-string for item 4, that's fine. They were asked to learn f-strings in the Session 1 homework, but some may not have internalized them yet.

## Part 2 — Dictionary practice

The learner creates a list of three patient dictionaries and prints a line for each.

````python
patients = [
    {"patient_id": "P001", "age": 58, "height_cm": 176.3, "weight_kg": 79.5},
    {"patient_id": "P005", "age": 46, "height_cm": 160.3, "weight_kg": 63.7},
    {"patient_id": "P012", "age": 71, "height_cm": 163.9, "weight_kg": 74.6},
]

print(f"{patients[0]['patient_id']} (age {patients[0]['age']}) has weight {patients[0]['weight_kg']} kg")
print(f"{patients[1]['patient_id']} (age {patients[1]['age']}) has weight {patients[1]['weight_kg']} kg")
print(f"{patients[2]['patient_id']} (age {patients[2]['age']}) has weight {patients[2]['weight_kg']} kg")
````

**Expected output:**
```
P001 (age 58) has weight 79.5 kg
P005 (age 46) has weight 63.7 kg
P012 (age 71) has weight 74.6 kg
```

**Instructor notes:**
- Some learners will create three separate dictionary variables instead of a list of dictionaries. That's acceptable — they haven't done "list of dicts" before. If they did use a list, that's great.
- Watch for quoting issues inside f-strings: `f"{patients[0]['patient_id']}"` requires mixing double and single quotes. Some learners will trip on this. A common workaround is assigning to a temp variable first.
- If the learner wrote three separate print statements without f-strings (e.g., `print(patients[0]["patient_id"], "(age", patients[0]["age"], ")...")`), that's fine. The output format matters more than the method.

## Part 6 — Reflection

Sample answers (learner responses will vary).

**Sample answers:**

1. *Pandas vs plain Python for mean:* With a list, I had to write `sum()/len()` myself. With pandas, I just called `.mean()`. Pandas also works directly on a column of thousands of values without me having to type them out.

2. *Single vs double brackets:* `df['age']` gives a Series, which is like a single column by itself. `df[['age']]` gives a DataFrame, which is still a table — it just happens to have one column. The difference matters because they have different methods and behaviors.

3. *read_csv vs read_sas:* Both load data into a DataFrame. `read_csv` is for CSV files (comma-separated text), `read_sas` is for XPT files (a format used by SAS/NHANES). Once loaded, the DataFrame works the same way regardless of where the data came from.

**Instructor notes:**
- For question 3, the key insight is that the file format is just how data is stored on disk. Once it's a DataFrame, the same operations apply. If the learner gets this, they understand a fundamental principle.
- If the reflection is thin or blank, raise it in the Session 3 warm-up. These reflections build the habit of self-assessment.

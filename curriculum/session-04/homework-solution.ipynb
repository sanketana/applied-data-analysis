## Part 1 — Warm-up: simple functions

````python
# 1a
def kg_to_lbs(weight_kg):
    return weight_kg * 2.205

print(kg_to_lbs(79.5))    # 175.3
````

````python
# 1b
def compute_bmi(weight_kg, height_cm):
    height_m = height_cm / 100
    return round(weight_kg / (height_m ** 2), 1)

print(compute_bmi(79.5, 176.3))    # 25.6
print(compute_bmi(47.7, 167.6))    # 17.0
print(compute_bmi(92.1, 157.7))    # 37.0
````

**Expected output:**
```
175.2975
25.6
17.0
37.0
```

**Instructor notes:**
- Check that both functions use `return`, not `print()`. If the learner used `print()` inside, the stall point from class hasn't landed. Revisit at the start of Session 5.
- `round()` is optional in 1b — learners might return the full decimal. That's fine. If they ask about rounding, show `round(value, 1)`.
- Some learners may forget to convert height from cm to m. If their BMI values are tiny (like 0.25), that's the cause.

## Part 2 — Build `classify_bmi()`

````python
def classify_bmi(weight_kg, height_cm):
    height_m = height_cm / 100
    bmi = weight_kg / (height_m ** 2)

    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"

print(classify_bmi(79.5, 176.3))    # Overweight
print(classify_bmi(47.7, 167.6))    # Underweight
print(classify_bmi(92.1, 157.7))    # Obese
````

**Expected output:**
```
Overweight
Underweight
Obese
```

**Instructor notes:**
- The classification logic is identical to Session 3 — the new skill is wrapping it in a function with `return`.
- If the learner hard-coded values (e.g., `bmi = 79.5 / (1.763 ** 2)`) instead of using the parameters, they don't understand what parameters are for. Ask: "What happens if you call this with a different patient?"
- Some learners may include `round()` on the BMI. That's fine — it doesn't affect the classification.

## Part 3 — Build `classify_bp()`

````python
def classify_bp(systolic):
    if systolic < 120:
        return "Normal"
    elif systolic < 130:
        return "Elevated"
    else:
        return "High"

print(f"Patient P001: systolic 119 → {classify_bp(119)}")
print(f"Patient P005: systolic 115 → {classify_bp(115)}")
print(f"Patient P007: systolic 143 → {classify_bp(143)}")
````

**Expected output:**
```
Patient P001: systolic 119 → Normal
Patient P005: systolic 115 → Normal
Patient P007: systolic 143 → High
```

**Instructor notes:**
- Learners will pick different patients. Check that the BP values come from the CSV and the classifications match.
- Watch for the parameter naming issue: if they define `classify_bp(systolic)` but write `if systolic_bp < 120` inside, they're using a variable name from Session 3 instead of the parameter. The NameError usually makes this clear.

## Part 5 — Apply to NHANES demographics

````python
demo = pd.read_sas("/content/drive/MyDrive/LEARNER-FOLDER/session-02/DEMO_J.XPT")
````

````python
# 5a
def nhanes_age_group(age):
    if age < 18:
        return "Minor"
    elif age < 45:
        return "Young adult"
    elif age < 65:
        return "Middle-aged"
    else:
        return "Senior"

# 5b
for i in range(10):
    age = demo['RIDAGEYR'][i]
    print(f"Respondent {i + 1}: age {int(age)} → {nhanes_age_group(age)}")
````

````python
# 5c
minors = 0
young_adults = 0
middle_aged = 0
seniors = 0

for i in range(100):
    group = nhanes_age_group(demo['RIDAGEYR'][i])
    if group == "Minor":
        minors += 1
    elif group == "Young adult":
        young_adults += 1
    elif group == "Middle-aged":
        middle_aged += 1
    else:
        seniors += 1

print(f"Minors: {minors}")
print(f"Young adults: {young_adults}")
print(f"Middle-aged: {middle_aged}")
print(f"Seniors: {seniors}")
````

**Expected output for 5b (approximate — depends on NHANES row order):**
```
Respondent 1: age 62 → Middle-aged
Respondent 2: age 53 → Middle-aged
...
```

**Expected output for 5c (approximate — first 100 respondents of NHANES 2017-2018):**
The distribution will vary, but expect a mix across all four groups. Minors are roughly 20–30% of NHANES respondents.

**Instructor notes:**
- Part 5c is the most complex code in this homework: it combines functions, loops, conditionals, and counters. If the learner got this working independently, they're ready for Session 5.
- The `int(age)` in 5b handles the fact that NHANES stores ages as floats (62.0 instead of 62). If the learner didn't use `int()`, the output looks slightly odd but isn't wrong.
- Some learners may try to use a dictionary to count groups instead of four separate variables. That's more advanced but perfectly valid — praise the instinct.

## Part 6 — Reflection

Sample answers (learner responses will vary).

**Sample answers:**

1. *Difference between print and return:* `print()` displays text on the screen but doesn't give anything back to the calling code. `return` sends a value back so you can store it in a variable or use it in further calculations. Without `return`, the function gives back `None`.

2. *When to write a function:* When you need to use the same logic more than once, or with different inputs. Instead of copying the if/elif block for every patient, you write it once as a function and call it each time.

3. *Rewriting Session 3 Part 1:* Instead of copying the BMI classification block three times, I'd write a `classify_bmi(weight_kg, height_cm)` function and call it three times: `classify_bmi(79.5, 176.3)`, `classify_bmi(73.2, 159.0)`, etc. Same results, much less code.

**Instructor notes:**
- For question 1, the key insight is the `None` part — a function without `return` doesn't crash, it silently gives `None`. If the learner mentions this, they've understood the stall point.
- For question 3, if they just say "use a function" without showing how, push for specifics: "What would the function's parameters be? What would it return?"
- If the reflection is thin, revisit in Session 5 warm-up.

## Part 7 — Stretch (optional)

````python
# 7a
def patient_summary(weight_kg, height_cm, systolic):
    bmi_cat = classify_bmi(weight_kg, height_cm)
    bp_cat = classify_bp(systolic)
    return f"BMI: {bmi_cat}, BP: {bp_cat}"

print(patient_summary(79.5, 176.3, 119))     # P001
print(patient_summary(92.1, 157.7, 149))     # P014
````

````python
# 7b
for i in range(len(df)):
    pid = df['patient_id'][i]
    age = df['age'][i]
    sex = df['sex'][i]
    summary = patient_summary(df['weight_kg'][i], df['height_cm'][i], df['systolic_bp'][i])
    print(f"{pid} (age {age}, {sex}): {summary}")
````

**Expected output for 7a:**
```
BMI: Overweight, BP: Normal
BMI: Obese, BP: High
```

**Instructor notes:**
- This stretch introduces functions calling other functions — a fundamental composition pattern. The learner doesn't need to know the term "composition," but seeing it work is valuable.
- In 7b, the output is the same as Part 4a, just organized differently. The point is that `patient_summary` bundles two function calls into one, making the loop cleaner.
- If the learner got this working, they have a good intuition for building up from small functions.

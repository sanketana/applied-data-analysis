## Part 1 — Quick recap: what a function does

````python
def cm_to_inches(height_cm):
    inches = height_cm * 0.3937
    return inches

print(cm_to_inches(176.3))
print(cm_to_inches(162.5))
````

A function is a named block of code that takes inputs (parameters), does some work, and gives back a result (`return`). You write it once, call it as many times as you want.

## Part 2 — The motivation: your Session 3 code

In the Session 3 homework, you wrote something like this three times:

````python
# Patient P001
height_m = 176.3 / 100
bmi = 79.5 / (height_m ** 2)

if bmi < 18.5:
    cat = "Underweight"
elif bmi < 25:
    cat = "Normal"
elif bmi < 30:
    cat = "Overweight"
else:
    cat = "Obese"

print(f"Patient P001: BMI = {bmi:.1f} → {cat}")
````

Three patients, three copies of the same block. The only things that changed were the weight and height values. Everything else was identical.

### Wrapping it in a function

````python
def classify_bmi(weight_kg, height_cm):
    height_m = height_cm / 100
    bmi = weight_kg / (height_m ** 2)

    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese" 
````

Now call it on three patients:

````python
print(classify_bmi(79.5, 176.3))   # P001
print(classify_bmi(47.7, 167.6))   # P015
print(classify_bmi(92.1, 157.7))   # P014
````

One function definition. Three calls. Same results as the copy-paste approach, but now you could classify 10,000 patients without writing any more classification code.

## Part 4 — The stall point: print vs return

Here's a version of `classify_bmi` that uses `print()` instead of `return`:

````python
def classify_bmi_buggy(weight_kg, height_cm):
    height_m = height_cm / 100
    bmi = weight_kg / (height_m ** 2)

    if bmi < 18.5:
        print("Underweight")
    elif bmi < 25:
        print("Normal")
    elif bmi < 30:
        print("Overweight")
    else:
        print("Obese")
````

It seems to work:

````python
classify_bmi_buggy(79.5, 176.3)
````

Looks fine — it printed "Overweight". But now try to store the result:

````python
result = classify_bmi_buggy(79.5, 176.3)
print(f"The category is: {result}")
````

Output:
```
Overweight
The category is: None
```

What happened?

- The function printed "Overweight" to the screen (that's the first line).
- But it didn't `return` anything, so `result` is `None`.
- `print()` talks to the screen. `return` talks to the code that called the function.

### The fix

Change every `print()` inside the function to `return`. Then the caller gets the actual value back:

````python
result = classify_bmi(79.5, 176.3)
print(f"The category is: {result}")
````

### The rule

> Use `return` when you need the function's result in your code. Use `print()` when you just want to display something.

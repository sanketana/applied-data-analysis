## The problem: repeating yourself

In the Session 3 homework, you classified three patients' BMI. Your code probably looked something like this:

````python
# Patient P001
height_m = 176.3 / 100
bmi = 79.5 / (height_m ** 2)

if bmi < 18.5:
    cat = "Underweight"
elif bmi < 25:
    cat = "Normal"
elif bmi < 30:
    cat = "Overweight"
else:
    cat = "Obese"

print(f"Patient P001: BMI = {bmi:.1f} → {cat}")
````

Functions solve this. You write the logic once, give it a name, and call it with different inputs.

## Defining a function with `def`

Here's a simple function that converts height from centimeters to inches:

````python
def cm_to_inches(height_cm):
    inches = height_cm * 0.3937
    return inches
````

The pieces:

- `def` tells Python you're defining a function.
- `cm_to_inches` is the name you choose (use snake_case, just like variable names).
- `height_cm` inside the parentheses is a **parameter** — a placeholder for the input value.
- The indented lines are the **function body** — the code that runs when you call the function.
- `return` sends the result back to whoever called the function.

---

## Calling a function

Defining a function doesn't run it. You need to **call** it by writing the name with parentheses and an input value:

````python
result = cm_to_inches(176.3)
print(result)
````

The value you pass in (176.3) is called an **argument**. It fills the `height_cm` parameter inside the function. The function runs, computes the result, and `return` sends it back. Now `result` holds 69.4.

You can call the same function with any input:

````python
print(cm_to_inches(162.5))
print(cm_to_inches(175.5))
print(cm_to_inches(157.7))
````

One function, many uses. No copy-pasting.

## Functions with multiple parameters

A function can take more than one input. Separate parameters with commas:

````python
def compute_bmi(weight_kg, height_cm):
    height_m = height_cm / 100
    bmi = weight_kg / (height_m ** 2)
    return round(bmi, 1)
````

````python
print(compute_bmi(79.5, 176.3))
print(compute_bmi(47.7, 167.6))
print(compute_bmi(92.1, 157.7))
````

The order of arguments matters. `compute_bmi(79.5, 176.3)` puts 79.5 into `weight_kg` and 176.3 into `height_cm`. Swapping them would give a wrong (and physiologically absurd) answer.

## `return` vs `print` — this distinction matters

Look at these two versions of a function:

````python
# Version A — with return
def classify_age(age):
    if age < 18:
        return "Minor"
    elif age < 65:
        return "Adult"
    else:
        return "Senior"

# Version B — with print
def classify_age_v2(age):
    if age < 18:
        print("Minor")
    elif age < 65:
        print("Adult")
    else:
        print("Senior")
````

Both seem to produce output when you call them. But watch what happens when you try to *use* the result:

````python
result_a = classify_age(45)
result_b = classify_age_v2(45)

print(f"Version A result: {result_a}")
print(f"Version B result: {result_b}")
````

`return` gives the value back to your code, so `result_a` is `"Adult"`.

`print()` puts text on the screen but gives back nothing, so `result_b` is `None`.

Rule of thumb: if your function computes a result that other code needs to use, use `return`.

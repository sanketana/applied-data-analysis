## Part 2 — Three patients (Block C exercise)

The learner computes BMI for three patients using variables.

```python
# Patient A
height_a = 1.70
weight_a = 68
bmi_a = weight_a / (height_a ** 2)
print(f"Patient A: BMI = {bmi_a:.1f}")
```

```python
# Patient B
height_b = 1.58
weight_b = 82
bmi_b = weight_b / (height_b ** 2)
print(f"Patient B: BMI = {bmi_b:.1f}")
```

```python
# Patient C
height_c = 1.82
weight_c = 75
bmi_c = weight_c / (height_c ** 2)
print(f"Patient C: BMI = {bmi_c:.1f}")
```

**Expected output:**
- Patient A: BMI = 23.5
- Patient B: BMI = 32.8
- Patient C: BMI = 22.6

**Instructor notes:**

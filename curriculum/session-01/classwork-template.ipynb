### Variables — labels that hold values

A variable is a label. We put a value next to the label with `=`, and we can use the label later.


```python
patient_age = 45
print(patient_age)
```

### The four basic data types you'll see constantly

- **Integer** (`int`) — whole numbers, like age
- **Float** (`float`) — numbers with decimals, like weight
- **String** (`str`) — text, like a name — always in quotes
- **Boolean** (`bool`) — `True` or `False`


```python
age = 45                  # integer
weight_kg = 72.4          # float
patient_name = "Priya"    # string
has_diabetes = False      # boolean

print(age, weight_kg, patient_name, has_diabetes)
```

### A first real calculation — BMI

BMI = weight in kg ÷ (height in metres)²


```python
height_m = 1.65
weight_kg = 72.4
bmi = weight_kg / (height_m ** 2)
print(bmi)
```

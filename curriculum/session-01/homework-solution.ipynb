## Part 1 — Variables practice

The learner creates five personal variables and prints them.

````python
my_name = "Priya"
my_age = 34
my_height_m = 1.63
my_weight_kg = 58.0
drinks_coffee = True

print(my_name)
print(my_age)
print(my_height_m)
print(my_weight_kg)
print(drinks_coffee)
````

**Instructor notes:**
- Values will differ — that's the point. Check that types are correct: string, int, float, float, bool.
- Common mistake: writing `drinks_coffee = "True"` (string instead of bool). Gently correct if you see quotes around True/False.
- Some learners will use `print(my_name, my_age, ...)` in one call. That's fine.
- If height is in cm (e.g., `163`) instead of metres, note it now — it matters for Part 2.

## Part 2 — Compute your BMI

The learner computes BMI from their Part 1 variables and prints it with an f-string.

````python
my_bmi = my_weight_kg / (my_height_m ** 2)
print(f"My BMI is {my_bmi:.1f}")
````

**Expected output (with sample values above):** `My BMI is 21.8`

**Instructor notes:**
- The f-string is new — the homework asks the learner to look it up. If they wrote `print("My BMI is", my_bmi)` instead, that's fine. Acknowledge the effort and show the f-string version briefly.
- Common mistake: forgetting parentheses around `height_m ** 2`, computing `weight / height ** 2` which Python interprets correctly due to operator precedence — but some learners write `(weight / height) ** 2`. Check the result is physiologically reasonable (18–40).
- If they used height in cm, the BMI will be astronomically large. Good teaching moment.

## Part 3 — Load the patient CSV and compute three BMIs

The learner loads the CSV, picks three patients, and computes BMI for each.

````python
import pandas as pd

csv_path = "/content/drive/MyDrive/LEARNER-FOLDER/session-01/patient_measurements_small.csv"
df = pd.read_csv(csv_path)
df.head(20)
````

````python
# Patient P001: height 176.3 cm, weight 79.5 kg
height_p001 = 176.3 / 100
weight_p001 = 79.5
bmi_p001 = weight_p001 / (height_p001 ** 2)
print(f"Patient P001: BMI = {bmi_p001:.1f}")

# Patient P005: height 160.3 cm, weight 63.7 kg
height_p005 = 160.3 / 100
weight_p005 = 63.7
bmi_p005 = weight_p005 / (height_p005 ** 2)
print(f"Patient P005: BMI = {bmi_p005:.1f}")

# Patient P012: height 163.9 cm, weight 74.6 kg
height_p012 = 163.9 / 100
weight_p012 = 74.6
bmi_p012 = weight_p012 / (height_p012 ** 2)
print(f"Patient P012: BMI = {bmi_p012:.1f}")
````

**Expected output:**
```
Patient P001: BMI = 25.6
Patient P005: BMI = 24.8
Patient P012: BMI = 27.8
```

**Instructor notes:**
- The learner picks any three patients — values will differ. Check that the results are physiologically reasonable.
- The CSV has height in **cm**, so the learner needs to convert to metres (divide by 100). This is a common stall point. If they forget, BMIs will be tiny fractions. Ask: "Does a BMI of 0.25 make sense for a person?"
- Some learners will try to use pandas indexing to extract values (e.g., `df.loc[0, 'height_cm']`). That's great if they figure it out, but the homework explicitly says hand-coding is fine.
- A few learners may try to loop through the DataFrame. Acknowledge the instinct but confirm that hand-coding three values is the intended approach for now.

## Part 4 — Reflection

Sample answers (learner responses will vary widely).

**Sample answers:**

1. *What is a variable?* — A variable is a name that holds a value so you can use it later. It's like labeling a container so you know what's inside.

2. *Why different types?* — Age is a whole number (you're 34, not 34.7), so Python stores it as an integer. Height has decimal precision, so it needs to be a float. The type tells Python how to store and work with the value.

3. *What surprised you about the CSV?* — Answers vary. Common observations: "I didn't expect so many columns," "The blood pressure values were higher than I expected," "Some patients are quite young."

**Instructor notes:**

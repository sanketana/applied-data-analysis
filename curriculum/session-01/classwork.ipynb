### Your very first line of code

### Variables — labels that hold values

A variable is a label. We put a value next to the label with `=`, and we can use the label later.

```python
patient_age = 45
print(patient_age)
```

### The four basic data types you'll see constantly

- **Integer** (`int`) — whole numbers, like age
- **Float** (`float`) — numbers with decimals, like weight
- **String** (`str`) — text, like a name — always in quotes
- **Boolean** (`bool`) — `True` or `False`

```python
age = 45                  # integer
weight_kg = 72.4          # float
patient_name = "Priya"    # string
has_diabetes = False      # boolean

print(age, weight_kg, patient_name, has_diabetes)
```

### A first real calculation — BMI

BMI = weight in kg ÷ (height in metres)²

```python
height_m = 1.65
weight_kg = 72.4
bmi = weight_kg / (height_m ** 2)
print(bmi)
```

## Part 2 — Your turn: three patients

Compute BMI for each of these patients and print the result:

| Patient | Height (m) | Weight (kg) |
|---|---|---|
| A | 1.70 | 68 |
| B | 1.58 | 82 |
| C | 1.82 | 75 |

```python
# Patient A
height_a = 1.70
weight_a = 68
bmi_a = weight_a / (height_a ** 2)
print(f"Patient A: BMI = {bmi_a:.1f}")
```

```python
# Patient B
height_b = 1.58
weight_b = 82
bmi_b = weight_b / (height_b ** 2)
print(f"Patient B: BMI = {bmi_b:.1f}")
```

```python
# Patient C
height_c = 1.82
weight_c = 75
bmi_c = weight_c / (height_c ** 2)
print(f"Patient C: BMI = {bmi_c:.1f}")
```
